### Notes

- Returns are daily, close to close, simple total returns. A weight set at the close of $t$ earns the return of $t+1$: $r^{\text{book}}_{t+1} = \boldsymbol{w}_t' \boldsymbol{r}_{t+1}$. Annualize with 252 trading days.

- VaR and CVaR are positive loss magnitudes, one-day at the 5% level unless a question says otherwise. A forecast for day $t$ uses data through $t-1$. A breach is a day whose realized loss exceeds the VaR forecast made for it. EWMA uses $\lambda = 0.94$ and a zero mean; you choose how to initialize $\sigma^2$ and state it.

- The **four-day loss** is the compounded loss of the daily-reset 4x book over July 24 to 29, 2026, $1 - \prod_{t=1}^{4}(1 + r_t)$, with $r_t$ the book's daily return.

```python
def path_stats(r):
    wealth = (1 + r).cumprod()
    dd = wealth / wealth.cummax() - 1
    trough = dd.idxmin()
    peak = wealth.loc[:trough].idxmax()

    return dd.min(), peak.date(), trough.date(), wealth.iloc[-1] - 1


def stats(r):
    ann_mean = r.mean() * 252
    ann_vol = r.std() * np.sqrt(252)
    sharpe = ann_mean / ann_vol
    skewness = r.skew()
    excess_kurt = r.kurt()
    
    worst_day = min(r)
    emp_VaR = -r.quantile(0.05)
    emp_CVaR = -r[r <= -emp_VaR].mean()
    
    max_drawdown, peak_date, trough_date, cum_return = path_stats(r)

    return pd.Series({
        'Ann. mean': ann_mean,
        'Ann. vol': ann_vol,
        'Sharpe': sharpe,
        'Skewness': skewness,
        'Excess Kurtosis': excess_kurt,
        "Worst Day": worst_day,
        "VaR": emp_VaR,
        "CVaR": emp_CVaR,
        "Max Drawdown": max_drawdown,
        "Peak": peak_date,
        "Trough": trough_date,
        "Cumulative Return": cum_return
    })

summary_stats = pd.DataFrame({"4x, 2021–2026": stats(book_4x[FULL]),
                            "4x, 2025–2026": stats(book_4x[RECENT]),
                            "1x, 2021–2026": stats(book_1x[FULL]),
                            "1x, 2025–2026": stats(book_1x[RECENT])})

pct_stats = ['Ann. mean', 'Ann. vol', 'Worst Day', 'VaR', 'CVaR', 'Max Drawdown', 'Cumulative Return']
ratio_stats = ['Sharpe', 'Skewness', 'Excess Kurtosis']

summary_stats.style.format('{:.2%}', subset=(pct_stats, summary_stats.columns)) \
                   .format('{:.2f}', subset=(ratio_stats, summary_stats.columns))
```

```python
num = summary_stats.drop(['Peak', 'Trough']).astype(float)
ratio = num['4x, 2021–2026'] / num['1x, 2021–2026']

print('Scales x4: ', list(ratio[np.isclose(ratio, 4)].index))
print('Invariant: ', list(ratio[np.isclose(ratio, 1)].index))
print('Non-linear:', list(ratio[~np.isclose(ratio, 4) & ~np.isclose(ratio, 1)].index))
```

**Answer 1.1.**

The descriptive statisticss respond to leverage in three different ways: they either scale by 4x, are invariant, or respond non-linearly. 

Annual mean, annual volatility, worst day, and empirical VaR and CVaR all scale linearly with leverage. This is expected, as they all depend on the size of each day's return, and all daily returns are scaled by 4x with leverage.

Sharpe, skewness and excess kurtosis remain equal across leverage levels. Sharpe remains the same because leverage amplifies returns but equally amplifies volatility. Skewness and excess kurtosis are characteristics of the daily return distribution which are unaffected by scaling, since they are standardized moments.

Finally, and most interestingly, maximum drawdown and cumulative returns are non-linear in leverage. Maximum drawdown goes from −26.3% to −79.4% (ratio 3.02) over 2021–2026. The reason it can't scale linearly because it is bounded by −100%, as the NAV cannot fall below 0, and because it is measured on the compounded path. Cumulative returns do not scale linearly because they are path-dependent and compound. Over 2021–2026 the 4x book returned 9,805% vs. 379% for the 1x book (ratio 25.9), and over 2025–2026 12,690% vs. 340% (ratio 37.3).

> For each window in 1.1, report the standard error of the 4x book's annualized mean and of its Sharpe ratio, and the Sharpe ratio's 95% band. You are reviewing the book on June 30, 2026 with the statistics from 1.1 to 1.4 in hand. Which window would you base a risk assessment on, and what is a Sharpe ratio estimated on eighteen months of daily data evidence of? Answer in a short paragraph that uses the standard errors.
>
> ### Estimation Error
> Under independence, the standard error of an annualized mean estimated from $N$ years of daily data is $\sigma_{\text{ann}}/\sqrt{N}$. With $q = 252$ periods a year, the annualized Sharpe ratio has standard error $\sqrt{(1 + \text{SR}^2/2q)/N}$.

```python
# 1.4
def estimation_error(r):
    N = len(r) / 252                                 
    ann_mean = r.mean() * 252
    ann_vol = r.std() * np.sqrt(252)
    sharpe = ann_mean / ann_vol

    se_mean = ann_vol / np.sqrt(N)
    se_sharpe = np.sqrt((1 + sharpe**2 / (2 * 252)) / N)

    return pd.Series({
        'Years (N)': N,
        'Ann. mean': ann_mean,
        'SE mean': se_mean,
        'Sharpe': sharpe,
        'SE Sharpe': se_sharpe,
        'Sharpe 95% low': sharpe - 1.96 * se_sharpe,
        'Sharpe 95% high': sharpe + 1.96 * se_sharpe,
    })

se_table = pd.DataFrame({'4x, 2021–2026': estimation_error(book_4x[FULL]),
                         '4x, 2025–2026': estimation_error(book_4x[RECENT])})
se_table.style.format('{:.2f}')
```

**Answer 1.4.**

We would base the risk assessment on the full window, as the standard errors for the mean and the Sharpe are around a third and half as large as the ones for the recent window (0.38 vs 1.12 for the mean and 0.43 vs 0.83 for the Sharpe, respectively). Furthermore, it includes 2021 to 2023, a regime in which the strategy lost money.

The Sharpe ratio estimated on the 18 months of data, 3.11, is evidence that the trade was working during that regime, but not necessarily an indication of the strategy's expected Sharpe. While it is high, its SE is also high (95% band of 1.48 to 4.74, vs 0.55 to 2.23 for the full window), and even higher if we consider the possibility that returns are not iid.

# 4. Where the Risk Sat

Section 4 uses daily excess returns and two estimation windows: 2021-01-04 to 2024-12-31 and 2025-01-02 to 2026-06-30. The 1x book keeps the betas readable; multiply by four for the 4x book.

```python
# Section 4 setup
EARLY = slice('2021-01-04', '2024-12-31')
WINDOWS = {'2021–2024': EARLY, '2025–2026': RECENT}
FACTOR_SETS = {'SPY': ['SPY'], 'SPY+SMH': ['SPY', 'SMH']}
Z_05 = 1.6448536269514722                      # Normal 5% quantile (scipy `stats` is shadowed by our 1.1 function)

excess = returns.sub(rf, axis=0)               # names and factors in excess of SHV
w_1x = weights / 4                             # +1/12 long, -1/12 short

sleeves_1x = pd.DataFrame({
    'Long sleeve':  excess[long_tickers]  @ w_1x[long_tickers],    # +50% of NAV
    'Short sleeve': excess[short_tickers] @ w_1x[short_tickers],   # -50% of NAV
    'Book 1x':      book_1x,                                       # already excess (weights sum to zero)
})


def regress(y, X):
    """OLS with intercept: alpha, betas, R², residuals."""
    Xc = np.column_stack([np.ones(len(X)), X.values])
    coef = np.linalg.lstsq(Xc, y.values, rcond=None)[0]
    resid = pd.Series(y.values - Xc @ coef, index=y.index)
    return coef[0], pd.Series(coef[1:], index=X.columns), 1 - resid.var() / y.var(), resid


def name_regressions(win, factors):
    """Regress each of the twelve names: B (k x N), alphas, residual panel."""
    X = excess.loc[win, factors]
    B, alphas, resid = {}, {}, {}
    for t in tickers:
        a, b, _, e = regress(excess.loc[win, t], X)
        B[t], alphas[t], resid[t] = b, a, e
    return pd.DataFrame(B), pd.Series(alphas), pd.DataFrame(resid)


models = {(wn, fn): name_regressions(win, fac)
          for wn, win in WINDOWS.items() for fn, fac in FACTOR_SETS.items()}
```

> The book has zero net exposure. What is its market beta in each window, and which sleeve betas produce it? Did the short sleeve hedge the long sleeve's largest exposure, add to it, or leave it alone?
>
> ### Portfolio Exposures
>
> Regress each name's excess return on the factor excess returns, $\tilde r_{i,t} = \alpha_i + \boldsymbol{\beta}_i' \tilde{\boldsymbol{f}}_t + \epsilon_{i,t}$, and stack the loadings in $B$, one column per name. The book with weights $\boldsymbol{w}$ has factor exposure $B\boldsymbol{w}$ and residual return $\boldsymbol{w}'\boldsymbol{\epsilon}_t$.

```python
#4.1
def factor_regression(y, X):
    a, b, r2, e = regress(y, X)
    out = {'Alpha (ann.)': a * 252}
    out.update({f'Beta {f}': b[f] for f in b.index})
    out.update({'R²': r2, 'Resid vol (ann.)': e.std() * np.sqrt(252)})
    return pd.Series(out)

reg_table = pd.DataFrame({
    (wn, fn, p): factor_regression(sleeves_1x.loc[win, p], excess.loc[win, fac])
    for wn, win in WINDOWS.items()
    for fn, fac in FACTOR_SETS.items()
    for p in sleeves_1x.columns
}).T
reg_table.index.names = ['Window', 'Factors', 'Portfolio']

reg_table.style.format('{:.2%}', subset=['Alpha (ann.)', 'Resid vol (ann.)']) \
               .format('{:.2f}', subset=[c for c in reg_table.columns if c.startswith('Beta') or c == 'R²'], na_rep='')
```

```python
# Verify b_p = B w: stacked name loadings vs. regressing the book directly
check = {}
for (wn, fn), (B, alphas, _) in models.items():
    b_stack = B[tickers] @ w_1x[tickers]
    _, b_direct, _, _ = regress(book_1x.loc[WINDOWS[wn]], excess.loc[WINDOWS[wn], FACTOR_SETS[fn]])
    for f in b_stack.index:
        check[(wn, fn, f)] = {'B w (1x)': b_stack[f], 'Direct (1x)': b_direct[f], 'Exposure (4x)': 4 * b_stack[f]}

exposure_check = pd.DataFrame(check).T
exposure_check.index.names = ['Window', 'Factors', 'Factor']
exposure_check.style.format('{:.3f}')
```

Stacking the twelve names' loadings and weighting them, b_p = B w, reproduces the book's directly estimated betas exactly in both windows and both factor sets, as it must by the linearity of OLS on a common sample.

The book has zero net dollars but not zero beta. On SPY alone, its market beta is 0.10 in 2021–2024 (0.40 at 4x) and 0.67 in 2025–2026 (2.69 at 4x). It comes from the gap between the sleeve betas: the long sleeve's +0.85 is almost fully offset by the short sleeve's −0.75 in 2021–2024, but in 2025–2026 the long sleeve's beta rises to +1.23 while the short sleeve's falls to −0.55. The long names' average SPY beta went from about 1.7 to 2.5, while the software names' fell from about 1.5 to 1.1, so equal dollars on each side no longer meant equal market exposure. The market model explains little of the book, though (R² of 0.01 and 0.12), and leaves an annualized alpha of 96% in 2025–2026, so SPY is not where most of the risk sat.

Adding SMH changes the picture. In 2025–2026 the long sleeve loads 0.74 on SMH and the short sleeve +0.15, so the book is +0.89 SMH and −0.97 SPY (+3.55 and −3.86 at 4x). The book's R² rises from 0.12 to 0.45, and the alpha falls from 96% to 57%. The positive SMH loading on the short sleeve means the software names had negative SMH betas once SPY was controlled for (ADBE −0.52, WDAY −0.61, CRM −0.43), so shorting them added semiconductor exposure. In 2021–2024 the short sleeve did slightly hedge the long sleeve's SMH loading (−0.10 against +0.32).

So the short sleeve hedged the market leg (SPY) but not the long sleeve's largest exposure. In 2025–2026 it added to the semiconductor exposure, and the book became a leveraged long-semiconductors, short-market position. Because SPY and SMH are highly correlated, the individual two-factor betas are imprecise, but their combination, long SMH against short SPY, is the robust part of the result.

## 4.2

> Estimate the 1x book's beta to `SPY` alone, and to `SPY` and `SMH` together, on a trailing one-year window at each quarter-end from September 2021 to June 2026. Tabulate or plot. The weights never changed. Did the exposures?

```python
#4.2
quarter_ends = pd.date_range('2021-09-30', '2026-06-30', freq='QE')   # use freq='Q' on pandas < 2.2

def rolling_betas(q, window=252):
    idx = book_1x.loc[:q].index[-window:]                              # trailing 252 days to the last trading day <= q
    _, b1, r2_1, _ = regress(book_1x.loc[idx], excess.loc[idx, ['SPY']])
    _, b2, r2_2, _ = regress(book_1x.loc[idx], excess.loc[idx, ['SPY', 'SMH']])
    return pd.Series({'SPY (SPY only)': b1['SPY'], 'R² (SPY only)': r2_1,
                      'SPY (SPY+SMH)': b2['SPY'], 'SMH (SPY+SMH)': b2['SMH'], 'R² (SPY+SMH)': r2_2})

rolling = pd.DataFrame({q.date(): rolling_betas(q) for q in quarter_ends}).T
rolling.index.name = 'Quarter end'

fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(rolling.index, rolling['SPY (SPY only)'], marker='o', label='SPY beta (SPY only)')
ax.plot(rolling.index, rolling['SPY (SPY+SMH)'], marker='s', label='SPY beta (SPY+SMH)')
ax.plot(rolling.index, rolling['SMH (SPY+SMH)'], marker='^', label='SMH beta (SPY+SMH)')
ax.axhline(0, color='black', lw=0.7)
ax.set_title('1x book: trailing one-year betas at quarter-ends')
ax.set_ylabel('Beta (1x)')
ax.legend()
plt.show()

rolling.style.format('{:.2f}')
```

The weights never changed, but the exposures did, and by a lot. From September 2021 to mid-2024 the book looked close to market-neutral. Its SPY-only beta stayed between −0.03 and 0.21 (essentially zero through 2022–2023), and SPY explained almost none of its variance (R² of 0.00 to 0.03). In the two-factor model, the SMH loading was small and stable at 0.15 to 0.21, offset by an SPY loading of −0.13 to −0.29. The two factors together explained under 12% of the variance.

From the second half of 2024 the exposures grew. The SPY-only beta jumped to 0.50 in September 2024 and held around 0.4 to 0.6 through 2025, before reaching 1.16 in June 2026 (4.6 at 4x). The SMH loading rose steadily, from 0.30 in September 2024 to 0.45 in March 2025, 0.64 at end-2025, then 0.92 and 0.98 in the first two quarters of 2026 (3.9 at 4x). The SPY loading in the two-factor model moved the other way, to −1.03. The two-factor R² rose from under 0.10 to 0.45, so the book was increasingly a long-semiconductors, short-market position rather than a stock-specific spread.

The largest increase came in the last two quarters before the reversal: the SMH loading roughly doubled between September 2025 and June 2026. The fixed-window betas in 4.1 understate where the book stood at the June 30 review (SMH 0.89 and SPY-only 0.67 over 2025–2026, against 0.98 and 1.16 on the trailing year). Two caveats apply. The trailing windows overlap, so adjacent quarters are not independent estimates. And the June 2026 SPY-only jump comes with an R² of only 0.15, so it is less reliable than the SMH trend. The SMH trend is steady across quarters and shows up in both factor sets.

A book with constant weights is therefore not a book with constant risk. As the long names became more semiconductor-like and the software shorts less so, the same positions turned into a much larger factor bet, without any trade being made.

## 4.3

> For each factor set in 4.1 and each window, report the average pairwise correlation of the residuals among the long names, among the short names, and across the sleeves, next to the same averages for the raw returns. Then, for the 2025-01-02 to 2026-06-30 window, compute the 1x book's variance three ways: from the sample covariance of the twelve names; from the factor model with a diagonal residual covariance; from the factor model with the full residual covariance. Do this for `SPY` alone and for `SPY` plus `SMH`.
>
> - Convert each variance to a 4x one-day Normal 5% VaR.
>
> After the factors are removed, how many dimensions of risk are left in this book? Which term does the diagonal drop, and how large is it as a share of the book's variance?
>
> ### Variance Decomposition
>
> $\sigma^2_{\text{book}} = (B\boldsymbol{w})'\,\Sigma_f\,(B\boldsymbol{w}) + \boldsymbol{w}'\,\Sigma_\epsilon\,\boldsymbol{w}$. A factor risk model usually replaces $\Sigma_\epsilon$ with its diagonal; the sample covariance of the twelve names needs no such assumption.

```python
#4.3
def block_corrs(df):
    return pd.Series({
        'Avg corr: long-long':   avg_corr(df[long_tickers]),
        'Avg corr: short-short': avg_corr(df[short_tickers]),
        'Avg corr: long-short':  df.corr().loc[long_tickers, short_tickers].values.mean(),
    })

resid_corrs = {}
for wn, win in WINDOWS.items():
    resid_corrs[(wn, 'Raw excess returns')] = block_corrs(excess.loc[win, tickers])
    for fn in FACTOR_SETS:
        resid_corrs[(wn, f'Residuals | {fn}')] = block_corrs(models[(wn, fn)][2])

resid_corr_table = pd.DataFrame(resid_corrs).T
resid_corr_table.index.names = ['Window', 'Series']
resid_corr_table.style.format('{:.2f}')
```

```python
w = w_1x[tickers]
Sigma = excess.loc[RECENT, tickers].cov()
var_sample = w @ Sigma @ w

var_rows = {('—', 'Sample covariance'): {'Variance (daily)': var_sample}}
for fn, fac in FACTOR_SETS.items():
    B, _, E = models[('2025–2026', fn)]
    b = B[tickers] @ w
    sys_var = b @ excess.loc[RECENT, fac].cov() @ b               # b' Σ_F b
    Sigma_e = E.cov()
    diag_var = (w**2 * np.diag(Sigma_e)).sum()                    # w' diag(Σ_ε) w
    full_var = w @ Sigma_e @ w                                    # w' Σ_ε w
    var_rows[(fn, 'Factor + diagonal residual')] = {'Variance (daily)': sys_var + diag_var, 'Systematic': sys_var,
                                                    'Residual diagonal': diag_var}
    var_rows[(fn, 'Factor + full residual')] = {'Variance (daily)': sys_var + full_var, 'Systematic': sys_var,
                                                'Residual diagonal': diag_var, 'Residual off-diagonal': full_var - diag_var}

var_table = pd.DataFrame(var_rows).T
var_table.index.names = ['Factors', 'Model']
var_table['Vol 1x (ann.)'] = np.sqrt(var_table['Variance (daily)'] * 252)
var_table['VaR 5% 4x (1-day)'] = Z_05 * 4 * np.sqrt(var_table['Variance (daily)'])
var_table['Systematic share'] = var_table['Systematic'] / var_sample
var_table['Off-diagonal share'] = var_table['Residual off-diagonal'] / var_sample
var_table['Missed by model'] = 1 - var_table['Variance (daily)'] / var_sample

var_table.style.format('{:.6f}', subset=['Variance (daily)', 'Systematic', 'Residual diagonal', 'Residual off-diagonal'], na_rep='') \
               .format('{:.2%}', subset=['Vol 1x (ann.)', 'VaR 5% 4x (1-day)', 'Systematic share',
                                         'Off-diagonal share', 'Missed by model'], na_rep='')
```

```python
# How many dimensions of residual risk are left? Eigenvalues of the residual correlation matrix, 2025–2026
eig_rows = {}
for fn in FACTOR_SETS:
    ev = np.sort(np.linalg.eigvalsh(models[('2025–2026', fn)][2].corr().values))[::-1]
    eig_rows[fn] = {'Eigenvalue 1': ev[0], 'Eigenvalue 2': ev[1], 'Eigenvalue 3': ev[2],
                    'PC1 share': ev[0] / ev.sum(),
                    'Effective dimensions': ev.sum()**2 / (ev**2).sum()}   # participation ratio, out of 12

eig_table = pd.DataFrame(eig_rows).T
eig_table.style.format('{:.2f}').format('{:.1%}', subset=['PC1 share'])
```

Over 2025-01-02 to 2026-06-30, the 1x book's sample variance is 0.000463 per day: 34.2% annualized volatility and a one-day Normal 5% VaR of 14.16% for the 4x book. With the full residual covariance, the factor model reproduces this exactly under both factor sets. It must, because in-sample OLS residuals are orthogonal to the factors. The diagonal model does not.

With SPY alone, the factor term explains only 12.4% of the book's variance and the diagonal residual 26.8%. The remaining 60.8% is the off-diagonal residual covariance, the term a diagonal risk model drops. The diagonal model therefore puts the 4x VaR at 8.86% instead of 14.16%, understating it by 37%. Adding SMH raises the systematic share to 44.6% and cuts the dropped term to 32.8% of the variance, but the diagonal VaR of 11.61% is still 18% below the sample figure.

The dropped term is Σ_{i≠j} w_i w_j Cov(ε_i, ε_j). Within each sleeve the weights have the same sign and across sleeves opposite signs. So positive residual correlation among the longs, positive residual correlation among the shorts, and negative residual correlation across sleeves all add to the book's variance rather than diversifying away. A diagonal model assumes all of this is zero, which is exactly the assumption a concentrated thematic long-short book violates.

After removing the factors, the residual correlation matrix is far from the identity. Under SPY its largest eigenvalue is 4.08 (34% of the total), followed by 2.33 and 1.08. Under SPY+SMH it is 3.34 (27.8%), 2.11 and 1.44. With 12 names and about 373 days, eigenvalues of pure noise would stay below roughly 1.4 (the Marchenko–Pastur upper edge, (1 + √(12/373))² ≈ 1.39). So about two dimensions of common residual risk are left under either factor set, plus a borderline third under SPY+SMH. The participation ratio gives 5.5 and 6.8 effective dimensions out of 12, well short of the twelve independent bets a diagonal model implicitly assumes. Adding SMH absorbs part of the first residual factor, but the book's risk still lies in a small number of themes — the hardware names together and the software names together — that neither SPY nor SMH fully captures.

## 4.4

> Attribute the July 24 to 29 loss of the 4x book to its factor exposure and its residual, as sums of daily contributions, using betas from each window and both factor sets from 4.1.
>
> - Report the systematic share under each of the four combinations.
>
> Was the reversal a factor event or a residual event, and does the answer depend on what you call a factor?

```python
#4.4
rev_4x = book_4x.loc[REVERSAL]

attribution = {}
for (wn, fn), (B, alphas, _) in models.items():
    fac = FACTOR_SETS[fn]
    b = B[tickers] @ w_1x[tickers]
    factor_daily = 4 * (excess.loc[REVERSAL, fac] @ b)           # 4x factor contribution each day
    alpha_daily = 4 * (alphas[tickers] @ w_1x[tickers]) * pd.Series(1.0, index=rev_4x.index)
    resid_daily = rev_4x - factor_daily - alpha_daily
    attribution[(wn, fn)] = {
        'Book (sum of daily)': rev_4x.sum(),
        'Factor': factor_daily.sum(),
        'Alpha': alpha_daily.sum(),
        'Residual': resid_daily.sum(),
        'Systematic share': factor_daily.sum() / rev_4x.sum(),
        **{f'4x beta {f}': 4 * b[f] for f in fac},
    }

attribution_table = pd.DataFrame(attribution).T
attribution_table.index.names = ['Beta window', 'Factors']

print('Compounded 4-day return, 4x:', f'{(1 + rev_4x).prod() - 1:.2%}')
print('Sum of daily returns, 4x:   ', f'{rev_4x.sum():.2%}')

attribution_table.style.format('{:.2%}', subset=['Book (sum of daily)', 'Factor', 'Alpha', 'Residual', 'Systematic share']) \
                       .format('{:.2f}', subset=[c for c in attribution_table.columns if c.startswith('4x beta')], na_rep='')
```

The daily-reset 4x book lost 66.9% compounded over July 24–29, and its daily returns sum to −96.4%. The attribution below splits the sum, since the factor and residual contributions add to the sum of daily returns, not to the compounded loss.

The systematic share depends almost entirely on the factor set and the beta window. With SPY alone it is negligible: 0.5% of the loss using 2021–2024 betas and 3.5% using 2025–2026 betas. That is because SPY barely moved during the reversal. Its excess return was +0.07%, +0.01% and +0.21% on the first three days and −1.57% on the last, −1.28% in total. A market-model risk system would have labeled almost the entire loss as residual. Adding SMH raises the systematic share to 11.5% with 2021–2024 betas and 46.0% with 2025–2026 betas. SMH fell on every day (−3.30%, −2.26%, −3.48%, −4.82%, −13.9% in total), and with the recent 4x loading of 3.55 that accounts for about −49% of the book's return. The −3.86 SPY loading gives back about +5%, because the book was effectively short the market and the market fell slightly. The intercept adds a few points of positive alpha in every case, so the residual is a little larger than the gap between the book's return and the factor term.

Even under the most favorable specification, more than half the loss (−55.7%) is residual, and it was not noise. The book lost between 19.6% and 27.4% on each of the four days while SMH fell only 2–5% a day. Measured against the 2025–2026 residual volatility at 4x (about 6.4% a day under SPY+SMH and 8.1% under SPY alone), the four-day residual is roughly a 4.3-sigma move under SPY+SMH and a 6.1-sigma move under SPY. That is very unlikely if residuals are independent across names, and 4.3 shows they are not. About two common residual dimensions remain after the factors are removed, and 61% (SPY) or 33% (SPY+SMH) of the book's variance sits in the off-diagonal residual covariance that a diagonal risk model drops.

So whether the reversal was a factor event or a residual event depends on what you call a factor. Relative to the market it was entirely a residual event. Relative to semiconductors it was partly a factor event, and only if you use betas from the boom period: the 2021–2024 betas, which a longer-history risk model would have relied on, attribute only 11.5% of the loss to the factors. The rest is the hardware-versus-software theme itself, a factor common to all twelve names that neither SPY nor SMH spans. If you added a theme factor (such as the long basket minus the short basket), nearly all of the loss would become systematic, but trivially so, since that factor is the book. The useful conclusion is that the loss was not idiosyncratic name risk that diversification would have reduced. It was concentrated exposure to one unhedged theme, which a factor model built on SPY, or even on SPY and SMH, would have classified as diversifiable residual.
